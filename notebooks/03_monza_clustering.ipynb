```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from f1_style_clustering.analysis.teammates import compare_teammates
from f1_style_clustering.config import load_circuit_config, load_config
from f1_style_clustering.data.loader import enable_cache, load_session
from f1_style_clustering.pipeline import run_pipeline
from f1_style_clustering.visualization.cluster_plots import plot_pca_clusters
from f1_style_clustering.visualization.radar_charts import plot_cluster_radar_chart
```

```python
cfg = load_config()
circuits = load_circuit_config()
enable_cache(cfg["general"]["cache_dir"])
session = load_session(2025, "Italian Grand Prix", "R")
out = run_pipeline(session, config=cfg, circuit_config=circuits)
print("k", out["k_used"], "rule", out["k_rule"], "zone_source", out["zone_source"])
cols = [c for c in ["Cluster", "Archetype", "Team", "n_laps"] if c in out["df_summary"].columns]
out["df_summary"][cols].sort_values("Cluster")
```
