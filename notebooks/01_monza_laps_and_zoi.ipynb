# 01 Monza — vueltas y ZOI

Carga la carrera de Monza 2025, filtra vueltas válidas y muestra las ventanas de frenado / vértice / salida sobre el trazado.


```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from f1_style_clustering.config import load_circuit_config, load_config
from f1_style_clustering.data.filters import filter_canonical_laps
from f1_style_clustering.data.loader import enable_cache, load_session
from f1_style_clustering.features.zoi import extract_circuit_zones, resolve_selected_corners
from f1_style_clustering.visualization.corner_maps import plot_circuit_with_zones
from f1_style_clustering.visualization.telemetry_plots import plot_corner_telemetry_comparison
```

```python
cfg = load_config()
circuits = load_circuit_config()
enable_cache(cfg["general"]["cache_dir"])
session = load_session(2025, "Italian Grand Prix", "R")
laps = filter_canonical_laps(session)
print(session.event.EventName, session.name, "drivers", len(session.drivers), "clean laps", len(laps))
```

Mapa: cada color es una ventana ZOI (frenado, vértice, salida). Ahí se recorta la telemetría de cada curva.


```python
selected, source = resolve_selected_corners(session, circuits)
zones = extract_circuit_zones(session, circuit_config=circuits, selected_corners=selected)
print("zone_source:", source)
for name, z in zones.items():
    print(name, z["typology"], "corners", z["corner_numbers"], "brake", z["braking"], "apex", z["apex"], "exit", z["exit"])
fig = plot_circuit_with_zones(session, analysis_zones=zones)
fig
```

Telemetría de un grupo de curvas (T1) partida en las tres ventanas. Compara dos o tres pilotos sobre el mismo recorte.


```python
t1 = zones["T1_Variante_del_Rettifilo"]
drivers = list(session.drivers[:3])
fig = plot_corner_telemetry_comparison(session, t1, corner_title="T1 Rettifilo", drivers=drivers)
fig
```
